```python
import pyautogui
from PIL import ImageGrab, ImageOps
import time

# 給自己 3 秒鐘切換到小恐龍遊戲畫面
print("請在 3 秒內切換到遊戲網頁...")
time.sleep(3)

def press_space():
    pyautogui.keyDown('space')
    time.sleep(0.05)                
    pyautogui.keyUp('space')
    print("跳！")

def start_game():
    # 這是偵測區的座標 (x, y, width, height)                                                     
    # 你需要根據你的螢幕調整這組數字    
    # 建議先截圖，確認小恐龍前方仙人掌會經過的位置
    box = (803, 292, 833, 315) 
    
    while True:   
        # 抓取偵測區的影像並轉為灰階                    
        image = ImageGrab.grab(box)
        gray_image = ImageOps.grayscale(image)
        
        # 計算區域內所有像素的總和
        # 如果背景是白色，當仙人掌（黑色）進入時，像素總和會改變
        pixel_sum = sum(gray_image.getdata())
        
        # 設定一個閾值，如果像素變化太大就跳
        # 這裡的數值需要根據實際偵測區的大小來測試
        if pixel_sum < 140000: # 假設背景全白時總和大於此值
            press_space()                
   
if __name__ == "__main__":
    start_game()
```
